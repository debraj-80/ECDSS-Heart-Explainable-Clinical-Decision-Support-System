## Apply Feature Engineering

```python
df_engineered = create_features(df)

new_features = [col for col in df_engineered.columns if col not in df.columns]
print(f'New engineered features: {new_features}')
print(f'Total features: {len(df_engineered.columns)}')
```

## Engineered Feature Descriptions

| Feature | Formula | Clinical Rationale |
|---------|---------|-------------------|
| RPP | thalach × trestbps | Rate Pressure Product - indicator of myocardial oxygen consumption |
| Thalach_Age_Ratio | thalach / age | Relative cardiac capacity normalized by age |
| Log_Chol | ln(chol + 1) | Reduces right-skew of cholesterol distribution |

**Note:** Pulse Pressure (true pulse pressure = systolic - diastolic) was **excluded** because the UCI dataset does not contain diastolic blood pressure measurements.
